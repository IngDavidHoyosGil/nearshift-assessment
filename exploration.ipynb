## items vs payments

```python
con.execute("""
    WITH items AS (
        SELECT order_id, COUNT(*) AS number_items
        FROM raw.olist_order_items_dataset
        GROUP BY order_id
    ),
    payments AS (
        SELECT order_id, COUNT(*) AS number_payments
        FROM raw.olist_order_payments_dataset
        GROUP BY order_id
    )
    SELECT
        i.order_id,
        i.number_items,
        p.number_payments,
        i.number_items * p.number_payments AS rows_after_merge
    FROM items i
    JOIN payments p USING (order_id)
    WHERE i.number_items > 1
    AND p.number_payments > 1
    LIMIT 20;
""").fetchdf()
```

## types, states and dates

```python
con.execute("""
    SELECT
        TRY_CAST(price AS DECIMAL(12, 2)) AS price_num
    FROM raw.olist_order_items_dataset
    LIMIT 10;
""").fetchdf()
```

```python
con.execute("""
    SELECT
        COUNT(*) AS rows,
        SUM(CASE WHEN price IS NULL THEN 1 ELSE 0 END) AS null_values,
        SUM(CASE
            WHEN price IS NOT NULL
            AND TRY_CAST(price AS DECIMAL(12, 2)) IS NULL
            THEN 1 ELSE 0
        END) AS values_no_convertibles
    FROM raw.olist_order_items_dataset;
""").fetchdf()
```

```python
con.execute("""
    SELECT order_status, COUNT(*) AS orders
    FROM raw.olist_orders_dataset
    GROUP BY order_status
    ORDER BY orders DESC;
""").fetchdf()
```

```python
con.execute("""
    SELECT
        COUNT(*) AS rows,
        SUM(CASE
            WHEN order_purchase_timestamp IS NOT NULL
            AND TRY_CAST(order_purchase_timestamp AS TIMESTAMP) IS NULL
            THEN 1 ELSE 0
        END) AS dates_no_convertibles,
        MIN(TRY_CAST(order_purchase_timestamp AS TIMESTAMP)) AS first_purchase,
        MAX(TRY_CAST(order_purchase_timestamp AS TIMESTAMP)) AS last_purchase
    FROM raw.olist_orders_dataset;
""").fetchdf()
```

## Model exploration

```python
con.execute("""
    SELECT
        'dim_customers' AS table_name,
        COUNT(*) AS total_rows,
        COUNT(DISTINCT customer_id) AS unique_keys
    FROM modeled.dim_customers

    UNION ALL

    SELECT
        'dim_products',
        COUNT(*),
        COUNT(DISTINCT product_id)
    FROM modeled.dim_products

    UNION ALL

    SELECT
        'fact_orders',
        COUNT(*),
        COUNT(DISTINCT order_id)
    FROM modeled.fact_orders;
""").fetchdf()
```

```python
con.execute("""  
SELECT
    'fact_order_items' AS table_name,
    COUNT(*) AS total_rows,
    COUNT(DISTINCT (order_id, order_item_id)) AS unique_keys
FROM modeled.fact_order_items

UNION ALL

SELECT
    'fact_payments',
    COUNT(*),
    COUNT(DISTINCT (order_id, payment_sequential))
FROM modeled.fact_payments;
""").fetchdf()
```

```python
con.execute("""
SELECT
    'fact_orders.order_purchase_timestamp' AS column_name,
    COUNT(*) FILTER (
        WHERE order_purchase_timestamp IS NULL
    ) AS null_rows
FROM modeled.fact_orders

UNION ALL

SELECT
    'fact_orders.order_delivered_customer_date',
    COUNT(*) FILTER (
        WHERE order_delivered_customer_date IS NULL
    )
FROM modeled.fact_orders

UNION ALL

SELECT
    'fact_order_items.price',
    COUNT(*) FILTER (WHERE price IS NULL)
FROM modeled.fact_order_items

UNION ALL

SELECT
    'fact_order_items.freight_value',
    COUNT(*) FILTER (WHERE freight_value IS NULL)
FROM modeled.fact_order_items

UNION ALL

SELECT
    'fact_payments.payment_value',
    COUNT(*) FILTER (WHERE payment_value IS NULL)
FROM modeled.fact_payments

UNION ALL

SELECT
    'dim_products.product_weight_g',
    COUNT(*) FILTER (WHERE product_weight_g IS NULL)
FROM modeled.dim_products;
""").fetchdf()
```

```python
con.execute("""
SELECT
    order_status,
    COUNT(*) AS orders_without_delivery_date
FROM modeled.fact_orders
WHERE order_delivered_customer_date IS NULL
GROUP BY order_status
ORDER BY orders_without_delivery_date DESC;
""").fetchdf()
```

```python
con.execute("""
SELECT
    order_id,
    order_status,
    order_purchase_timestamp,
    order_delivered_carrier_date,
    order_delivered_customer_date,
    order_estimated_delivery_date
FROM modeled.fact_orders
WHERE
    (order_status = 'delivered'
        AND order_delivered_customer_date IS NULL)
    OR
    (order_status = 'canceled'
        AND order_delivered_customer_date IS NOT NULL)
ORDER BY order_status, order_id;
""").fetchdf()
```

```python
con.execute("""
SELECT
    'orders_without_customer' AS check_name,
    COUNT(*) AS issue_count
FROM modeled.fact_orders o
LEFT JOIN modeled.dim_customers c
    ON o.customer_id = c.customer_id
WHERE c.customer_id IS NULL

UNION ALL

SELECT
    'items_without_order',
    COUNT(*)
FROM modeled.fact_order_items i
LEFT JOIN modeled.fact_orders o
    ON i.order_id = o.order_id
WHERE o.order_id IS NULL

UNION ALL

SELECT
    'items_without_product',
    COUNT(*)
FROM modeled.fact_order_items i
LEFT JOIN modeled.dim_products p
    ON i.product_id = p.product_id
WHERE p.product_id IS NULL

UNION ALL

SELECT
    'payments_without_order',
    COUNT(*)
FROM modeled.fact_payments p
LEFT JOIN modeled.fact_orders o
    ON p.order_id = o.order_id
WHERE o.order_id IS NULL;
""").fetchdf()
```
